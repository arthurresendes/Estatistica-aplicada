```python
import numpy as np
import pandas as pd
```

```python
dados1 = dados.drop(columns=['CO_MUNICIPIO_RESIDENCIA'])
dados1.head()
```

```python
dados1 = dados1.drop(columns=['CO_UF_RESIDENCIA', 'SG_UF_RESIDENCIA', 'CO_MUNICIPIO_NASCIMENTO'])
dados1.head()
```

```python
dados1 = dados1.drop(columns=['NO_MUNICIPIO_RESIDENCIA', 'CO_UF_NASCIMENTO','SG_UF_NASCIMENTO', 'TP_ANO_CONCLUIU', 'TP_ENSINO', 'CO_MUNICIPIO_ESC' 
                                ,'CO_MUNICIPIO_ESC', 'CO_UF_ESC', 'SG_UF_ESC'],)
dados1.head()
```

```python
dados1 = dados1.drop(columns=['TP_DEPENDENCIA_ADM_ESC', 'TP_LOCALIZACAO_ESC', 'TP_SIT_FUNC_ESC'],)
dados1.head()
```

```python
dados1.loc[:, 'NU_NOTA_CN'] /= 10
```

```python
dados1.head()
```

```python
dados1.loc[:, 'NU_NOTA_CH'] /= 10
dados1.loc[:, 'NU_NOTA_MT'] /= 10
dados1.loc[:, 'NU_NOTA_LC'] /= 10
```

```python
dados1.head()
```

```python
dados1 = dados1.rename(columns={'NU_NOTA_REDACAO': 'NOTA_REDACAO'})
```

```python
dados1 = dados1.rename(columns={'NU_NOTA_CN': 'NOTA_CN', 'NU_NOTA_MT': 'NOTA_MT', 'NU_NOTA_CH': 'NOTA_CH',
                        'NU_NOTA_LC': 'NOTA_LC'})
```

```python
dados1 = dados1.rename(columns={'NU_IDADE': 'IDADE', 'TP_SEXO': 'SEXO', 'TP_COR_RACA': 'RACA', 'Q025': 'INTERNET', 'TP_ESCOLA': 'ESCOLA'
                        ,'NU_NOTA_COMP1': 'COMP1','NU_NOTA_COMP2': 'COMP2','NU_NOTA_COMP3': 'COMP3','NU_NOTA_COMP4': 'COMP4','NU_NOTA_COMP5': 'COMP5' })
```

```python
dados1['RACA'] = dados1['RACA'].replace({0: 'nao_declarado', 1: 'branco', 2: 'preta', 3: 'parda', 4: 'amarela', 5: 'indigena'})
```

```python
dados1['TP_LINGUA'] = dados1['TP_LINGUA'].replace([0,1], ['Inglês', 'espanhol'])
```

```python
dados1.head()
```

```python
dados1['ESCOLA'] = dados1['ESCOLA'].replace([1,2,3,4], ['Não respondeu', 'Pública', 'Privada', 'Exterior'])
dados1['INTERNET'] = dados1['INTERNET'].replace({'A': 'Não', 'B': 'Sim'})
```
