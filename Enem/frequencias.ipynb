```python
import pandas as pd
import numpy as np
from collections import Counter
```

```python
enem_sp = pd.read_csv("enem_tratado.csv" , encoding="iso-8859-1")
```

```python
enem_sp.head()
```

```python
enem_sorocaba_nascimento = enem_sp.loc[enem_sp.NO_MUNICIPIO_NASCIMENTO == "Sorocaba"]
```

```python
enem_sorocaba_nascimento.head()
```

```python
idade = enem_sorocaba_nascimento["IDADE"]
```

```python
idade
```

```python
freq_absoluta = Counter(idade)
```

```python
freq_absoluta
```

```python
freq_absoluta = pd.DataFrame.from_dict(freq_absoluta, orient='index')
```

```python
freq_absoluta
```

```python
freq_absoluta = freq_absoluta.sort_index(ascending=True)
freq_absoluta
```

```python
freq_relativa = freq_absoluta / freq_absoluta.sum()
freq_relativa
```

```python
freq_relativa_perc = freq_relativa * 100
freq_relativa_perc
```
