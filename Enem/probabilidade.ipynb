```python
def probabilidade(A,E):
    resultado = (A/E) * 100
    print('{:.2f}'.format(resultado))
```

```python
def probabilidade_nao(A,E):
    resultado = (1 - (A/E)) * 100
    print('{:.2f}'.format(resultado))
```

```python
probabilidade(len(enem_guarulhos), len(enem_sp))
```

```python
probabilidade(len(enem_bauru), len(enem_sp))
```

```python
probabilidade_nao(len(enem_guarulhos), len(enem_sp))
```

```python
probabilidade_nao(len(enem_bauru), len(enem_sp))
```

```python
def probabilidade_uniao(A,B,E):
    resultado = (A/E + B/E) * 100
    print('{:.2f}'.format(resultado))
```

```python
probabilidade_uniao(len(enem_guarulhos), len(enem_bauru), len(enem_sp))
```

```python
def probabilidade_intersec(A,B,E):
    resultado = (A/E * B/E) * 100
    print('{:.3f}'.format(resultado))
```

```python
probabilidade_intersec(len(enem_guarulhos), len(enem_bauru), len(enem_sp))
```

```python
def probabilidade_condi(AB,B):
    resultado = (AB / B) * 100
    print('{:.3f}'.format(resultado))
```
