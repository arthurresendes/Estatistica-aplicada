```python
area_quadrada = lambda x: x**2
print(area_quadrada(5))
```

```python
lista = [0,2,4,6,8,10,11,12,13]
mapeando = list(map(lambda x: x*2, lista))
print(mapeando)
```

```python
print([i*5 for i in lista])
print([i*5 for i in lista if i%2==0])
```

```python
conceito = ['Bom' if i >= 8 else 'Ruim' for i in lista]
print(conceito)
```
